```python
food1 = {'Khinkali','Mwvadi' ,'Khachapuri','Soko'}
food2 = {'Khinkali','Salati','Achma'}
khorci = food1.intersection(food2)
print(khorci)
```

```python
food1 = {'Khinkali','Mwvadi' ,'Khachapuri','Soko'}
food2 = {'Khinkali','Salati','Achma'}
food1.symmetric_difference_update(food2) #using symetric difference to change food1 to anti intersection items
print(food2)
```

```python
food1 = {'Khinkali','Mwvadi' ,'Khachapuri','Soko'}
food2 = {'Khinkali','Salati','Achma'}
vegan = food1.symmetric_difference(food2) # creates new set that is using symmetric difference (anti intersection)
print(vegan)
```

```python
result = food1.isdisjoint(food2) # checks if  they have joint section or not
print(result)
```

```python
result = food1.issubset(food2)
print(result)
```
